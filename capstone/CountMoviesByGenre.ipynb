# Import Required Libraries
Import the necessary libraries, including pandas and matplotlib.

```python
# Import the necessary libraries, including pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt
```

# Load Dataset
Load the movie dataset into a pandas DataFrame.

```python
# Load the movie dataset into a pandas DataFrame
df = pd.read_csv('ml-32m/movies.csv')  # Replace 'movies.csv' with the path to your dataset

# Display the first few rows of the DataFrame to verify loading
df.head()
```

# Data Preprocessing
Clean and preprocess the data, including handling missing values and extracting genres.

```python
# Data Preprocessing

# Check for missing values in the dataset
missing_values = df.isnull().sum()
print("Missing values in each column:\n", missing_values)

# Drop rows with missing values
df.dropna(inplace=True)

# Extract genres from the 'genres' column
# Assuming the 'genres' column contains a string of genres separated by '|'
df['genres'] = df['genres'].str.split('|')

# Explode the 'genres' column to have one genre per row
df = df.explode('genres')

# Display the first few rows of the DataFrame after preprocessing
df.head()
```

# Count Movies by Genre
Count the number of movies in each genre using pandas.

```python
# Count Movies by Genre

# Count the number of movies in each genre
genre_counts = df['genres'].value_counts()

# Display the counts
print("Number of movies in each genre:\n", genre_counts)

# Plot the counts as a bar plot
plt.figure(figsize=(10, 6))
genre_counts.plot(kind='bar')
plt.title('Number of Movies by Genre')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

# Plot Bar Graph
Use matplotlib to create a bar plot of the counts of movies by genre.

```python
# Plot Bar Graph
# Use matplotlib to create a bar plot of the counts of movies by genre.

plt.figure(figsize=(10, 6))  # Set the figure size
genre_counts.plot(kind='bar', color='skyblue')  # Plot the bar graph with skyblue color
plt.title('Number of Movies by Genre')  # Set the title of the plot
plt.xlabel('Genre')  # Set the x-axis label
plt.ylabel('Count')  # Set the y-axis label
plt.xticks(rotation=45)  # Rotate the x-axis labels for better readability
plt.show()  # Display the plot
```
