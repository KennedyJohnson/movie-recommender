# Import Required Libraries
Import the necessary libraries, including pandas and matplotlib.

```python
# Import the necessary libraries, including pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt
```

# Load Dataset
Load the movie ratings dataset into a pandas DataFrame.

```python
# Load the movie ratings dataset into a pandas DataFrame
ratings_df = pd.read_csv('ml-32m/ratings.csv')

# Display the first few rows of the DataFrame to ensure it loaded correctly
ratings_df.head()
```

# Count Ratings
Count the number of occurrences of each rating in the dataset.

```python
# Count Ratings
# Count the number of occurrences of each rating in the dataset
rating_counts = ratings_df['rating'].value_counts()

# Create a bar plot of the rating counts
plt.figure(figsize=(10, 6))
rating_counts.plot(kind='bar', color='skyblue')
plt.title('Count of Ratings Given')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```
